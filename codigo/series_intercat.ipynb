```python
import pandas as pd

df_ind = pd.read_csv('/home/camilo/figuras_mauro/data/cleaned_indexes.csv')

df_biofonia = pd.read_csv('/home/camilo/figuras_mauro/data/cleaned_soundscape.csv')
# renombrar columna Biophony -> Biofonia si existe
if 'Biophony' in df_biofonia.columns:
    df_biofonia.rename(columns={'Biophony': 'Biofonia'}, inplace=True)

df_riqueza = pd.read_csv('/home/camilo/figuras_mauro/data/riqueza_especies.csv')
# renombrar columnas en df_riqueza y asegurar tipo datetime
df_riqueza.rename(columns={'datetime_5min': 'datetime', 'species_count': 'Riqueza'}, inplace=True)
```

```python
import pandas as pd
import plotly.express as px

window_size = 72

# --- Ejemplo de unión de varios DF (ajusta nombres/rutas) ---

# df con índices acústicos
df_ind = df_ind.copy()
df_ind['datetime'] = pd.to_datetime(df_ind['datetime'], errors='coerce')
df_ind = df_ind.dropna(subset=['datetime'])

# df_biofonia
df_biofonia = df_biofonia.copy()
df_biofonia['datetime'] = pd.to_datetime(df_biofonia['datetime'], errors='coerce')
df_biofonia = df_biofonia.dropna(subset=['datetime'])

# df_riqueza
df_riqueza = df_riqueza.copy()
df_riqueza['datetime'] = pd.to_datetime(df_riqueza['datetime'], errors='coerce')
df_riqueza = df_riqueza.dropna(subset=['datetime'])

# Combinar por datetime y site (left para mantener todos los registros de df_ind)
df_all = (
    df_ind
    .merge(df_biofonia[['datetime', 'site', 'Biofonia']],
           on=['datetime', 'site'], how='left')
    .merge(df_riqueza[['datetime', 'site', 'Riqueza']],
           on=['datetime', 'site'], how='left')
)

# Lista de índices disponibles en el DF combinado
indices_disponibles = [
    'Biofonia', 'Riqueza', 'Ht', 'Hf', 'H',
    'ACI', 'NDSI', 'ADI', 'AEI', 'BI',
    'MEANt', 'VARt', 'MEANf', 'VARf', 'SKEWf', 'KURTf'
]

# --- Función general para graficar cualquier índice del DF combinado ---

def graficar_indice_plotly(df_comb, indice, window_size=72):
    if indice not in df_comb.columns:
        raise ValueError(f"El índice '{indice}' no existe en el DataFrame combinado.")

    df_plot = df_comb[['datetime', 'site', indice]].dropna(subset=[indice]).copy()
    df_plot = df_plot.sort_values('datetime')

    # suavizado por sitio
    df_plot['smoothed'] = (
        df_plot
        .groupby('site')[indice]
        .transform(lambda s: s.rolling(window=window_size,
                                      min_periods=1,
                                      center=True).mean())
    )

    fig = px.line(
        df_plot,
        x='datetime',
        y='smoothed',
        color='site',
        labels={'smoothed': indice, 'datetime': 'Fecha', 'site': 'Site'},
        title=f'Serie temporal suavizada: {indice}'
    )

    fig.update_xaxes(dtick="D1", tickformat="%m-%d", title_text="Fecha")
    fig.update_layout(legend_title_text="Sites")
    fig.show()

# --- Uso: cambiar el índice que quieres ver ---

indice = 'Biofonia'
graficar_indice_plotly(df_all, indice, window_size)

# Para otro:
# indice = 'Riqueza'
# graficar_indice_plotly(df_all, indice, window_size)
# indice = 'Ht'
# graficar_indice_plotly(df_all, indice, window_size)
```
