```python
import numpy as np
import matplotlib.pyplot as plt

# Datos de ejemplo
horas = np.linspace(0, 24, 25)
tiempo_espacio = np.linspace(1, 20, 20)
H, T = np.meshgrid(horas, tiempo_espacio)

# Superficie simulada
complejidad_sonora = (
    np.sin(H / 4) +
    np.cos(T / 6) +
    np.random.normal(0, 0.1, H.shape)
)

# Crear figura 3D
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Superficie
surf = ax.plot_surface(
    H, T, complejidad_sonora,
    cmap='viridis',
    linewidth=0,
    antialiased=True
)

# Etiquetas en español
ax.set_xlabel('Horas')
ax.set_ylabel('Tiempo/Espacio')
ax.set_zlabel('Índice de Complejidad Sonora')

# Barra de color
cbar = fig.colorbar(surf, shrink=0.6)
cbar.set_label('Índice de Complejidad Sonora')

# Título
ax.set_title('Variación de la Complejidad Sonora')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Ejes
horas = np.linspace(0, 23, 100)
tiempo_espacio = np.linspace(1, 20, 30)
H, T = np.meshgrid(horas, tiempo_espacio)

# Patrón diario uniforme (misma forma para todos)
patron_diario = np.sin(2 * np.pi * H / 24)

# Superficie: mismo patrón + pequeño ruido
complejidad_uniforme = (
    patron_diario +
    np.random.normal(0, 0.05, H.shape)
)

# Gráfica
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(
    H, T, complejidad_uniforme,
    cmap='viridis',
    linewidth=0
)

ax.set_xlabel('Horas')
ax.set_ylabel('Tiempo/Espacio')
ax.set_zlabel('Índice Acústico')
#ax.set_title('Patrón Diario Uniforme en el Tiempo/Espacio')

cbar = fig.colorbar(surf, shrink=0.6)
cbar.set_label('')

plt.show()
```

```python
# Variación del patrón según Tiempo/Espacio
fase_variable = T / 5
amplitud_variable = 0.5 + T / 25

complejidad_no_uniforme = (
    amplitud_variable *
    np.sin(2 * np.pi * H / 24 + fase_variable) +
    np.random.normal(0, 0.05, H.shape)
)

# Gráfica
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(
    H, T, complejidad_no_uniforme,
    cmap='viridis',
    linewidth=0
)

ax.set_xlabel('Horas')
ax.set_ylabel('Tiempo/Espacio')
ax.set_zlabel('Índice de Complejidad Sonora')
ax.set_title('Patrón Diario Variable en el Tiempo/Espacio')

cbar = fig.colorbar(surf, shrink=0.6)
cbar.set_label('Índice de Complejidad Sonora')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Ejes
horas = np.linspace(0, 23, 100)
tiempo_espacio = np.linspace(1, 20, 30)
H, T = np.meshgrid(horas, tiempo_espacio)

# Ruido base (blanco)
ruido_blanco = np.random.normal(0, 1, H.shape)

# Suavizado ligero en horas (para evitar ruido pixelado)
ruido_suavizado = np.zeros_like(ruido_blanco)
for i in range(len(tiempo_espacio)):
    ruido_suavizado[i] = np.convolve(
        ruido_blanco[i],
        np.ones(5)/5,
        mode='same'
    )

# Modulación estocástica dependiente de Tiempo/Espacio
modulacion = np.random.uniform(0.5, 1.5, (len(tiempo_espacio), 1))

# Superficie estocástica
complejidad_estocastica = (
    modulacion * ruido_suavizado
)

# Gráfica
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(
    H, T, complejidad_estocastica,
    cmap='viridis',
    linewidth=0,
    antialiased=True
)

ax.set_xlabel('Horas')
ax.set_ylabel('Tiempo/Espacio')
ax.set_zlabel('Índice Acústico')
#ax.set_title('Variación Horaria Estocástica en el Tiempo/Espacio')

cbar = fig.colorbar(surf, shrink=0.6)
cbar.set_label('')

plt.show()
```
